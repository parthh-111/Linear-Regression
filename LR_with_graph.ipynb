```python
from sklearn.linear_model import LinearRegression 
model = LinearRegression()
x = [[1],[2],[3],[4]]
y = [2,4,6,8]
model.fit(x,y)
prediction = model.predict([[3]])
print("Prediction for input is : " , prediction)
```

## linear regression by user input

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
x = [[1],[2],[3],[4]]
y = [2,4,6,8]
model.fit(x,y)
# take user input
user_value = float(input("enter a number to predict : "))
#predict value 
prediction = model.predict([[user_value]])
print("prediction is : " , prediction[0])
```

```python
import matplotlib.pyplot as plt
plt.scatter(x, y, color='blue', label='Data Points')
plt.plot(x, model.predict(x), color='orange', label='Regression Line')
plt.legend(loc='center left', bbox_to_anchor=(1, 1))
plt.title("Linear Regression Line")
plt.xlabel("X")
plt.ylabel("Y")
plt.tight_layout()
plt.show()
```
